```python
import requests, pandas, warnings

warnings.filterwarnings('ignore')

import matplotlib.pyplot as plt

plt.style.use('seaborn-v0_8-whitegrid')
```

```python
urljson = requests.get("https://storage.googleapis.com/dqlab-dataset/update.json")

urljson.status_code
```

```python
updates = rurljson['update']

len(updates)
```

```python
updates.keys()
```

```python
updates['total'].keys()
```

```python
digitalisasi = lambda i : "{:,}".format(i).replace(',', '.')
```

```python
total_positif = updates['total']['jumlah_positif']

print(digitalisasi(total_positif))
```

```python
total_meninggal = updates['total']['jumlah_meninggal']

print(digitalisasi(total_meninggal))
```

```python
total_sembuh = updates['total']['jumlah_sembuh']

print(digitalisasi(total_sembuh))
```

```python
total_hospitalized = updates['total']['jumlah_dirawat']

print(digitalisasi(total_hospitalized))
```

```python
fatalityRate = round((total_meninggal / total_positif) * 100, 3)

fatalityRate
```

```python
total_content = len(updates['harian'])

total_content
```

```python
columns = [i for i in updates['harian'][0].keys() if i.startswith('jumlah_') or i.startswith('key_')]

content = pandas.DataFrame(data = updates['harian'], columns = columns)
```

```python
content = content.applymap(lambda x: x['value'] if isinstance(x, dict) else x)

content['key_as_string'] = pandas.to_datetime(content['key_as_string']).dt.date

content = content.rename(columns = {'key_as_string': 'Tanggal'})

content = content.set_index('Tanggal')
```

```python
daily = content.iloc[:, :4]

cummulative = content.iloc[:, 4:]

del content
```

```python
daily.iloc[:5]
```

```python
fig, ax = plt.subplots(figsize = (12, 6))

ax.plot(cummulative['jumlah_positif_kum'])

ax.set_title("Jumlah Positif Kumulatif Pandemi Covid 19 di Indonesia")

ax.set_xlabel("Tanggal")

ax.set_ylabel("Jumlah Positif");
```

```python
fig, ax = plt.subplots(figsize = (12, 6))

ax.plot(cummulative['jumlah_meninggal_kum'])

ax.set_title("Jumlah Positif Pasien Meninggal Pandemi Covid 19 di Indonesia")

ax.set_xlabel("Tanggal")

ax.set_ylabel("Jumlah Meninggal");
```
